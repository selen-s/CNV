```python
# Create an ESKF class for single bearing estimation based in quaternions
import numpy as np 
class ESKF:
    def __init__(self, x0, P0):
        self.x = x0 # initial state estimate
        self.P = P0 # initial state covariance

    def predict(self, F, Q):
        self.x = F @ self.x # predict the next state
        self.P = F @ self.P @ F.T + Q # predict the next state covariance

    def update(self, H, R, z):
        y = z - H @ self.x  # measurement innovation
        S = H @ self.P @ H.T + R  # innovation covariance
        K = self.P @ H.T @ np.linalg.inv(S)  # Kalman gain
        self.x = self.x + K @ y  # update state estimate
        I_KH = np.eye(self.x.size) - K @ H  # identity minus Kalman gain times observation matrix
        self.P = I_KH @ self.P @ I_KH.T + K @ R @ K.T  # update state covariance

    def quaternion_normalize(self, q):
        return q / np.linalg.norm(q)  # normalize the quaternion to ensure it remains a valid rotation


```
